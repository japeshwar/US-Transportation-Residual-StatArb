```python
fig, axes = plt.subplots(1, 6, figsize = (16, 5))
dates = ['2016-01-01', '2018-01-01', '2020-01-01', '2022-01-01','2024-01-01', '2026-01-01']
for ax, d in zip(axes, dates):
    subset = stocks_returns[airline_returns.index <= d].tail(252)
    corr = subset.corr()
    im = ax.imshow(corr.values, cmap='RdYlGn', vmin = 0, vmax = 1.0)
    ax.set_xticks(range(len(corr))); ax.set_xticklabels(corr.columns, rotation = 45, fontsize = 10)
    ax.set_yticks(range(len(corr))); ax.set_yticklabels(corr.columns, fontsize = 12)
    ax.set_title(f"Correlation\nThrough {d[:10]}")
plt.suptitle("How Transportation Correlations Change Over Time")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize = (16, 4))

macro_returns['SPY'].cumsum().plot(ax = axes[0], title = 'SPY Cumulative Log Return')
macro_returns['Brent'].plot(ax = axes[1], title = 'Brent Price Level')
macro_returns['VIX'].plot(ax = axes[2], title = 'VIX Level')

plt.tight_layout()
plt.show()
```
